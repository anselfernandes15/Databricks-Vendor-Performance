```python
s3_bucket =  f's3://vendor-performance-1511/data/'
catalog_name = 'vendor_performance'
schema_name = 'gold'
files_data = dbutils.fs.ls(s3_bucket)
```

```python
for file in files_data:
    # 1. Read the CSV from S3
    df = (spark.read
        .format("csv")
        .option("header", "true")
        .option("inferSchema", "true")
        .load(f"{s3_bucket}{file.name}"))
    # 2. Write as a Delta Table (Managed Table)
    df.write.format("delta").mode("overwrite").saveAsTable(f"{catalog_name}.{schema_name}.{file.name[:-4]}")
```
